# Results and uncertainty

This notebook loads per-question outputs and computes paper-cluster bootstrap intervals. It deliberately reports no placeholders when experiments have not been run.

```python
from pathlib import Path

import pandas as pd

from scientific_rag.statistics import cluster_bootstrap_ci

primary = Path("../results/experiments/full_validation")
methods = ("bm25", "dense", "hybrid", "hybrid_reranked")
files = [
    primary / method / "evaluation/token-aware/retrieval_metrics.csv"
    for method in methods
    if (primary / method / "evaluation/token-aware/retrieval_metrics.csv").exists()
]
if files:
    for path in files:
        frame = pd.read_csv(path)
        print(
            path.parents[2].name,
            cluster_bootstrap_ci(frame.to_dict("records"), "evidence_recall"),
        )
else:
    print("Run the full-validation experiments first.")
```
