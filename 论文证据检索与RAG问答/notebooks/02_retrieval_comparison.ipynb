# Retrieval comparison

Run retrieval methods with `scripts/run_retrieval.py`, then load only the actual CSV outputs here.

```python
from pathlib import Path

import pandas as pd

primary = Path("../results/experiments/full_validation")
methods = ("bm25", "dense", "hybrid", "hybrid_reranked")
files = [
    primary / method / "evaluation/token-aware/retrieval_metrics.csv"
    for method in methods
    if (primary / method / "evaluation/token-aware/retrieval_metrics.csv").exists()
]
metrics = pd.concat([pd.read_csv(path) for path in files], ignore_index=True) if files else pd.DataFrame()
metrics.groupby("retriever")[
    ["evidence_recall", "mrr", "hit_rate", "complete_evidence_coverage"]
].agg(["mean", "count"]) if not metrics.empty else "Run the full-validation experiments first."
```
